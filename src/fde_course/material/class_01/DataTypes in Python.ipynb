### DataTypes in Python

Numeric : int, float, complex

text : str

Sequence : list, tuple, set, frozenset, range

mapping : dict ({key:value})

boolean : bool (True, False)

other datatypes : bytes, bytearray, memoryview, None

str     = Ordered, Immutable, Duplicaates are allowed        >>  ' ', " ", ''' ''', """ """
list    = Ordered, Mutable, Duplicaates are allowed          >> [] 
tuple   = Ordered, Immutable, Duplicaates are allowed        >> ()

set     = Unordered, Mutable, Duplicaates are not allowed    >> {}, set()    


```python
int1 = 45
print(type(int1), int1)
```

```python
float1 = 4.50
print(type(float1), float1)
```

```python
complex1 = 4.50 + 9j
print(type(complex1), complex1)
```

```python
str1 = 'Python'
print(type(str1), str1)
```

```python
l1 = [1,2,3.14,'Data', 2+3j]
print(type(l1), l1)
```

```python
t1 = (1,2,3.14,'Data', 2+3j)
print(type(t1), t1)
```

```python
s1 = {1,2,3.14,'Data', 2+3j}
print(type(s1), s1)
```

```python
fs1 = frozenset({1,2,3.14,'Data', 2+3j})
print(type(fs1), fs1)
```

```python
b1 = True
print(type(b1), b1)
```

```python
b1 = False
print(type(b1), b1)
```

```python
d1 = {'Virat':18, 'Rohit':45}  # key : value
print(type(d1), d1)
```

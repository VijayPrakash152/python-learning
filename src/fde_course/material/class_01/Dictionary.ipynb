### Dictionary

```python
Dict : { key1 : value1, key2 : value2 }

1. It is a collection of key and value pairs
2. Duplicate key is not allowed
3. Duplicate values are allowed
4. Comma seperated
5. Dict is enclosed by {}
6. We cannot use indexing
7. Mutable

key    >> Immutable datatyes >> int, float, complex, str, tuple, frozenset
value  >> Any datatyes       >> int, float, complex, str, list, tuple, set, frozenset, dict, array, df etc.

if Python  >= 3.7  >> Ordered 
if Python  <  3.7  >> Unordered


It is majorly used for mapping
'Rohit' to be mapped with 45
'Rohit' to be mapped with 'India'
'ABD' to be mapped with 'RSA'
```

### Sorting of dict using values

```python
sub_marks = {'phy': 87, 'bio': 85, 'che': 78, 'math': 89, 'eng': 81}

output = {}
for i in sorted(sub_marks.values()):   #i >> values >> marks
    for j in sub_marks:   # j >> keys >> subjects
        if i == sub_marks[j]:
            output[j] = i

output
```

```python
sub_marks = {'phy': 87, 'bio': 85, 'che': 78, 'math': 89, 'eng': 81}

output = {}
for i in sorted(sub_marks.values()):   #i >> values >> marks
    for j in sub_marks:   # j >> keys >> subjects
        if i == sub_marks[j]:
            output[j] = sub_marks[j]

output
```

```python
sub_marks = {'phy': 87, 'bio': 85, 'che': 78, 'math': 89, 'eng': 81}

output = {}
for i in reversed(sorted(sub_marks.values())):   #i >> values >> marks
    for j in sub_marks:   # j >> keys >> subjects
        if i == sub_marks[j]:
            output[j] = sub_marks[j]

output
```

```python
sub_marks = {'phy': 87, 'bio': 85, 'che': 78, 'math': 89, 'eng': 81}

output = {}
for i in sorted(sub_marks.values(),reverse=True):   #i >> values >> marks
    for j in sub_marks:   # j >> keys >> subjects
        if i == sub_marks[j]:
            output[j] = sub_marks[j]

output
```

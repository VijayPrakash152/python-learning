### Type Casting

##### 1. List to tuple 

```python
x = [4,5,7,8,9]
print(x)
y = tuple(x)
y
```

##### 2. Tuple to list

```python
x = (4,5,7,8,9)
print(x)
y = list(x)
y
```

```python
x
```

```python
x = (4,5,7,8,9)
print(x)

y = list(x)
y.append(10)
print(y)

x = tuple(y)
x
```

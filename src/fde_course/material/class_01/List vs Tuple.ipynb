### Tuple

```python
1. Immutable
2. Ordered
3. Duplicates are allowed
4. Tuples are enclosed by ()
5. Tuples are less functional
6. Tuples are faster tha lists
7. Tuples takes less memory due to immutability
8. It is a collection of heterogenous datatypes
9. Tuple object can be used as a key in dictionary
10. List inside tuple is mutable:
    x = (2,3,4,[5,6,7], 6,7,8)
    We can modify the contents of the list without violating the tuple immmutability
```

### List

```python
1. Mutable
2. Ordered
3. Duplicates are allowed
4. Lists are enclosed by []
5. Lists are more functional
6. Lists are slower tha tuples
7. Lists takes more memory due to mutability
8. It is a collection of heterogenous datatypes
9. Lists object can not be used as a key in dictionary
```

| Feature             | List                            | Tuple                                    |
| ------------------- | ------------------------------- | ---------------------------------------- |
| Definition          | Ordered collection of elements  | Ordered collection of elements           |
| Mutability          | Mutable                         | Immutable                                |
| Can modify elements | Yes                             | No                                       |
| Add elements        | Yes                             | No                                       |
| Remove elements     | Yes                             | No                                       |
| Duplicate elements  | Allowed                         | Allowed                                  |
| Order               | Maintains order                 | Maintains order                          |
| Indexing            | Supported                       | Supported                                |
| Slicing             | Supported                       | Supported                                |
| Hashable            | No                              | Yes, if all elements are hashable        |
| Dictionary key      | Cannot be used                  | Can be used if all elements are hashable |
| Syntax              | `[1, 2, 3]`                     | `(1, 2, 3)`                              |
| Performance         | Generally slightly slower       | Generally slightly faster                |
| Memory usage        | Generally higher                | Generally lower                          |
| Main use            | Data that may need modification | Data that should remain unchanged        |


| Feature                | Set                                   | Frozenset                               |
| ---------------------- | ------------------------------------- | --------------------------------------- |
| Definition             | Mutable collection of unique elements | Immutable collection of unique elements |
| Mutability             | Mutable                               | Immutable                               |
| Duplicate elements     | Not allowed                           | Not allowed                             |
| Order                  | Unordered                             | Unordered                               |
| Indexing               | Not supported                         | Not supported                           |
| Add elements           | `add()` supported                     | Not supported                           |
| Remove elements        | `remove()`, `discard()` supported     | Not supported                           |
| Update elements        | `update()` supported                  | Not supported                           |
| Hashable               | No                                    | Yes                                     |
| Dictionary key         | Cannot be used                        | Can be used                             |
| Element of another set | Cannot be used                        | Can be used                             |
| Syntax                 | `{1, 2, 3}`                           | `frozenset([1, 2, 3])`                  |
| Main use               | When data needs to be modified        | When data must remain unchanged         |

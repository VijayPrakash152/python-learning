List
A List is an ordered collection of elements.
It is mutable, meaning values can be changed.
Lists are written using square brackets [].
It can store different data types together.
Example: my_list = [10, "Python", 3.5]

### Deep Copy

While dealing with nested list, 2D-array

```python
import copy
```

```python
x = [[2,3,4], 'python',[10,20,30],[50,60,70], 'science',[100,200,300]]
y = copy.deepcopy(x) # deep copy

x[3][-1] = 700

print('x :', x)
print('y :', y)
```

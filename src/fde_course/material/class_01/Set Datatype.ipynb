### How to access set items

```python
s1 = {5,4,8,7,3,9}
s1
```

```python
s1[2]
```

```python
s1[2:4]
```

```python
s1 = {5,4,8,7,3,9, 21}

for i in s1:
    print(i)
```

```python
s1 = {5,4,8,'Python',7,3,9, 21, }

for i in s1:
    print(i)
```

```python
list1 = [3,4,5,6,7]
list2 = [5,6,1,2,10]

for i in list1:
    if i in list2:
        print(i)
```

```python
list1 = [3,4,5,6,7]
list2 = [5,6,1,2,10]
list3 = []
for i in list1:
    if i in list2:
        list3.append(i)

list3
```

```python
list1 = [3,4,5,6,7]
list2 = [5,6,1,2,10]
list3 = []

for i in list1:
    if i not in list2:
        list3.append(i)

list3
```

```python
list1 = [3,4,5,6,7]
list2 = [5,6,1,2,10]
list3 = []

for i in list2:
    if i not in list1:
        list3.append(i)

list3
```

```python
list1 = [3,4,5,6,7]
list2 = [5,6,1,2,10]

s1 = set(list1)
s1.intersection(list2)
```

```python
s1.difference(list2)
```

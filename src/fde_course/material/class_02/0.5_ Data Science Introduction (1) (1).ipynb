```python
It is a process of extracting meaningful information from data using Scientific Methods

Data should be in numeric form for all data science algorithm(ML,DL,NLP,etc.)
ex. Image >> Numeric
    Video >> Numeric
    Text  >> Numeric
    Audio >> Numeric

Dataformat for models >> Numpy Array or Dataframe
```

### Data Science Project Steps

```python
1. Problem Statement (Use Case/Business Problem)
2. Data Gathering(Data Collection)
3. Exploratory Data Analysis(EDA)
4. Feature Engineering
5. Feature Selection/Feature Extraction
6. Model Training(ML, DL,NLP)
7. Model Evaluation(Classification, Regression)
8. Create API by using Web Framework(Flask, Fast API, DJango) and Testing on Postman
9. Project Deployement on Cloud Platforms(AWS, Azure, GCP, Herrokku-Free)
```

### Types of Variables

```python
A column in your excel.csv, DB:
    Feature, Variables, Parameters, Columns

1. Independent Variables:
    Input Column:
        Categorical
        Contineous(Numeric)

2. Dependent Variable:
    Target Column:
        Categorical:(Classification)
            Binary:
                Yes/No, Spam/not spam, 0/1, Approved/Declined
            Multiclass: (More than 2 classes)
                Adhar/PAN/DL, Petrol/Electric/Diesel/Hybrid
        Contineous:(Regression)
            Price
            Salary
            Sales
            Amount
            Production
            Score
            Temprature
```

## 2. Data Science Project Steps

1. Problem Statement/Business Problem (Business Understanding) >> POC >> Proof of Concept
2. Data Gathering(DataBase)
3. Exploratory Data Analysis(EDA) >> Pandas, Numpy, Seaborn, Matplotlib
4. Feature Engineering/Data Pre-Processing:
    1. Scaling:
        Normalization
        Standardization
    2. Feature Transformation:
        Log, Squareroot, Cuberoot, BoxCox
    3. Encoding:
        1. One Hot
        2. Label 
    4. Handling Outlier
    5. Binning (Cont to Cat)
    
5. Feature Selection and Feature Extraction: (Dimensionality Reduction)
    Best set of features
    Feature Selection:
        1. Filter
        2. Wrapper
        3. Embedded 
    Feature Extraction:
        1. PCA
        2. LDA

6. Model Training/Building:
    1. Linear Regression
    2. KNN Regressor
    3. DT Regressor
    4. RF Regressor
    
7. Evaluation:
    Regression:
        1. MSE
        2. RMSE
        3. MAE
        4. R-Squared Value
        
    Classification:
        1. Classification Report(Precision, Recall,F1-Score, Wtd Avg, Mcr Avg)
        2. Confusion Matrix
        3. AUC-ROC Curve
        4. Accuracy
        
8. Model Tuning/Optimization:
    Hyperparamters Tuning

9. Web Development Framework: (API) >> Python Developer
    1. Flask
    2. Django
    3. Fast API
    4. gRPC
    
    These are used to create APIs
    Test APIs on Postman Application

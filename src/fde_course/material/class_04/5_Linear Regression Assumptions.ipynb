## Linear Regression Assumptions

Assumptions/Limitations: (based on Error term /Distribution of Data)
    
1) Linearity:
    >> Model assumes that there is linear relationship between independent and dependent var
    
2) Independence:
    >> Model assumes that the independent var are independent of each other
    
3) No Multicolinearity:
    >> Ind var should not be correlated with ind var
    >> if there is high correlation between ind vars, 
       So it becomes difficult to find the relation between predictor and target var

4) Homoscedasticity(Same scatter of error/variance):
    >> The spread of errors across the model should remain constant

5) Normality of Error Distribution:
    >> The errors should be normally distributed across the model

6) No Autocorrelation of Errors:
    >> The errors should not have correlation,  should not have any similar pattern
    
7) Mean of Residuals is zero:
    >> Beacause we are trying to get BFL exactly in the middle of points/residuals
    >> Residuals gets segregated into positive and negative, they migh cancel out each other

    ** Variance Inflation Factor (VIF) >> It Detects Multicolinearity

- VIF ranges from 1 onwards
- VIF = 1 >> NOt Correlated
- 1 to 5 >> Less Correlated (if we want to keep less features)
- 5 to 10 >> (Slightly Correlated features) if we want to keep more features
- more than 10 >> Highly Correlated

Loss Function/Cost Function >> It is a parameter, we try to minimize or optimise in gradient descent

    So all the metrics are basically cost functions itself:
        
    3) Mean Absolute Error(MAE) : It is an absolute diff between act and pred (MAPE : Mean Absolute Percentage Error)
        - MAE = sum((|act - pred|)/n)
        - MAE is robust to outliers
        
    4) Mean Squared Error(MSE) : 
        - MSE = sum((act-pred)**2)/n
        - MOst commonly used cost function
        - It tells us about performance of the model
        - Penelises the errors
        - It is sensitive to outliers
        
    5) Root Mean Squared Error(RMSE):
        RMSE = sqrt(sum((act-pred)**2)/n)
        
        - It is not robust compare to MAE
        - Preffered for Deep LEarning/Time Series

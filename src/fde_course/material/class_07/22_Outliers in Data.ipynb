Out of the box values

Datapoints which are far away observed values

Smaller or Larger values than remaining values

## How to Detect Outliers

1. Boxplot(graphical method-IQR)
2. Scatterplot
3. Z-Score(Normally Distributed Data)
4. IQR Method
5. Isolation Forest
6. kdeplot

## How to Handle Outliers

1. Delete Observations/Trimming: (Not Prefereble)
2. Imputation/Capping:
    1. Median
    2. Mean
    3. Zero
    4. Upper Limit
    5. Lower Limit
    6. Any static value
    
3. Transformations:(Used to reduce the impact of Outliers)
    Primarily it is used to handle skewness:
    1. Log Transformations
    2. Squareroot Transformations
    3. Cube root Transformations
    4. Reciprocal Transformations
    5. Box-COx Transformations
    6. Standardization
    7. Robust Scaling

## Impact of Outliers

1. Reduces the power of statistical analysis
2. High impact on mean and std values
3. Algorithm will not perform well in the presence of Outliers
4. Impact on basic assumptions of regression(Normality, Hetroscedasticty)

## Impact of Outliers in Algorithm

#### Sensitive to Outliers

Linear & Distance-Based Algo :->
1. Linear Regresssion
2. Logistic Regresssion
3. K-Nearest Neighbor
4. Support Vector Machines
5. K-Means Clustering

#### Not Sensitive to Outliers

CART Algo :->
1. Decision Tree
2. Random Forest

4. XGBoost

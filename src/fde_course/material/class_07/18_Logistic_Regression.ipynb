## Logistic Regression

>>> It is a Supervised ML algo which predicts categorical values based on probabilities of target class/event.

>> It is Classification Algorithm.
>> It use the same concept of Linear Regression
>> It is a probabilistic algo (0 to 1)
>> It is a parametric algo

Target Feature >> Discrete/Categorical:
    Binary Classification:
        Target Variable consist of 2 Values:
            True or False
            Spam or Not-Spam
            Defaulter or Non-Defaulter
            Diabetic or Non-Diabetic
            0 or 1
    
    Multi-Class Classification:
        Low-Medium-High
        0-1-2-3
        PAN-ADHAR-PASSPORT
        
        One vs Rest (OVR)
        One vs All
        
        Low vs (Medium and High)
        Medium vs (Low and High)
        High vs (Low and Medium)

## Advantages

1. Makes no assumption on distribution of residual
2. We can easily use binary and multi-class classification.
3. Quick to train
4. Very fast at classifying unknown records
5. Good accuracy on many dataset
6. Resisting to overfitting

## Disadvantages

1. Sensitive to outliers
2. Requires a large dataset
3. Requires linearly seperable data,Constructs linear boundary
4. Does not work well with overlapping classes

## Assumptions

>> Mainly 3
1. Target variable should be categorical in nature(binary, ordinal, multilass)
2. Constructs linear boundary
3. Requires large dataset

4. No Multicolinearity
5. Independence

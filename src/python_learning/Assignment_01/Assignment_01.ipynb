## 1.  What is a List?

A list is a mutable heterogenous ordered data structure in python that allows duplicates. For example - 

```python
my_list = [1,2,3,4,5,4]
heterogeneous_list = [1, "two", 3.0, True, None]
my_list.append(6) #mutable
print(my_list)
print(heterogeneous_list)
print(my_list[0:-2])
```

## 2.  What is a Tuple? 

A tuple is an ordered, heterogenous, immutable data structure in python that allows duplicates. For example - 

```python
my_tuple = (1,2,3,4,5,6)
heterogeneous_tuple = (1,"two", 3.0, True, None)
my_tuple = my_tuple + (7,8) #immutable
print(my_tuple)
print(heterogeneous_tuple)
```

## 3.  What is the difference between List and Tuple? 

List is a mutable data structure while tuple is an immutable data structure. This means that you can change the contents of a list after it has been created, while you cannot change the contents of a tuple after it has been created.
List is denoted by []
Tuple is denoted by ()

```python
my_list = [1,2,3,4,5]
my_tuple = (1,2,3,4,5)

my_list[0] = 10 #mutable
print(my_list)
my_tuple[0] = 10 #immutable, error will be raised
print(my_tuple)
```

## 12. Python program to find the smallest and largest number in a list (Without min-max function)

```python
import math

my_list = [1,2,3,4,5,6]

def maxElem(my_list):
    max_elem = -math.inf
    for num in my_list:
        if num > max_elem:
            max_elem = num

    return max_elem

def minElem(my_list):
    min_elem = math.inf
    for num in my_list:
        if num < min_elem:
            min_elem = num

    return min_elem     

print(f"The max element in the list {my_list} is {maxElem(my_list=my_list)}")  
print(f"The min element in the list {my_list} is {minElem(my_list=my_list)}")  
```

## 15. You are given a list of integer elements. Make a new list that will store a square of elements
## of the previous list. (With and without list comprehension)
## ● Input_list = [2,5,6,12]
## ● Output_list = [4,25,36,144]

```python
input_list = [2,5,6,12]

## Without list comprehension
output_list = []

for num in input_list:
    output_list.append(num*num)

print(f"The output list without list comprehension is {output_list}")


## With list comprehension 

output_list_lc = [elem*elem for elem in input_list]
print(f"The output list with list comprehension is {output_list_lc}");
```

## 26. How to flatten a list in Python?

```python
my_list = [1,2,3,[4,5,6], [7,[8,9]]];

def flatten_list(input_list):
    flattened_list = []
    for elem in input_list:
        if isinstance(elem, list):
            flattened_list.extend(flatten_list(elem))
        else:
            flattened_list.append(elem)

    return flattened_list

print(f"The flattened list is {flatten_list(my_list)}")


```

## 27. How to sort a list in ascending and descending order without using the sort function?

```python
my_list = [8,3,1,2,0,5,4, 9,7,6]

def bubble_sort(list_to_sort, ascending = True):
    n = len(list_to_sort)
    for i in range(n):
        for j in range(0, n-i-1):
            to_swap = list_to_sort[j] > list_to_sort[j + 1] if ascending == True else list_to_sort[j] < list_to_sort[j + 1]
            if(to_swap):
                list_to_sort[j], list_to_sort[j+1] = list_to_sort[j+1], list_to_sort[j]

bubble_sort(my_list,ascending = False)
print(f"The sorted list is descending order is - {my_list}")
bubble_sort(my_list,ascending = True)
print(f"The sorted list is ascending order is - {my_list}")


```

## 35. What is the difference between copy and deep copy?

```python
from copy import deepcopy

my_list = [1,2,3,4,5,[6,7,8,9]]

copied_list  =  my_list.copy() ## returns a shallow copy of the list
deep_copied_list = deepcopy(my_list) ## returns a deep copy of the list
my_list[5].append(10)

print(copied_list) ## A shallow copy creates a new outer list, but nested objects are still shared, so when we make changes in nested list in my_list, it will also reflect in copied_list.
print(deep_copied_list) ## A deep copy creates a new outer list and recursively copies all nested objects, so changes in my_list will not affect deep_copied_list.
```

## 1. What is a dictionary in Python? List out all properties of Dictionary. 

A dictionary in Python is a mutable data structure that stores data as key-value pairs.

A **dictionary** in Python is a mutable data structure that stores data as **key-value pairs**.

```python
student = {
    "name": "Vijay",
    "age": 25,
    "course": "Python"
}
```

### Properties of a Dictionary

1. Stores data as **key-value pairs**.
2. Keys must be **unique**.
3. Keys must be **immutable and hashable** types, such as strings, numbers, or tuples.
4. Values can be of **any data type**.
5. Dictionaries are **mutable**, so items can be added, changed, or removed.
6. Dictionaries preserve **insertion order** in modern Python versions.
7. Dictionaries can contain **different data types**.
8. Dictionaries can be **nested** inside other dictionaries.
9. Values may be duplicated, but keys cannot.
10. Dictionaries are **dynamically sized**.
11. They support fast lookup using keys, typically with average $O(1)$ time complexity.
12. An empty dictionary is created using `{}` or `dict()`.
13. Accessing a missing key with `dictionary[key]` raises a `KeyError`; `.get()` can avoid this.
14. Dictionaries are iterable, and iteration normally proceeds over their keys.

## 2. Explain the concept of dictionary comprehension in Python. 

````python
## Dictionary comprehension is a way to construct dictionary using a for loop in a single line.
my_list = [1,2,3,4,5]
my_dict = { elem: elem**2 for elem in my_list} ## elem becomes key and sqaure of elem becomes the value
print(my_dict)
````

## 3. Explain the difference between the dict.get(key) method and accessing dict[key] directly. 

````python
my_dict = {1:2, 2: 3, 3:4}

print(my_dict.get(5)) ## does not gives error if key is not present 
print(my_dict[5]) ## throws error if key is not present
````

## 4. How can you merge two dictionaries in Python? 

````python
dict_1 = {"x": 1, "y":2, "z":3}
dict_2 = {"a": 1, "b":2, "c":3}

dict_3 = {**dict_1, **dict_2}

print(dict_3)

## another approach

dict_4 = dict_1 | dict_2
print(dict_4)
````

## 5. How can you update dictionary values in Python? Provide examples. 

````python
my_dict = {"x": 1, "y":2, "z":3}

# my_dict.update({"x": 2}); one method , use it when we have to update mulitple key, values at one time.
my_dict["x"] = 2 # used when we have to update single value 
print(my_dict)
````

## 6.How can you iterate over a dictionary to access keys, values, and key-value pairs?

````python
my_dict = {"x": 1, "y":2, "z":3}

## printing keys
for key in my_dict:
    print(key)

## printing values
for value in my_dict.values():
    print(value)

## printing key,value pairs 
for key,value in my_dict.items():
    print(f"{key} : {value}")
````

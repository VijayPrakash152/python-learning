## 1. What is a dictionary in Python? List out all properties of Dictionary. 

A dictionary in Python is a mutable data structure that stores data as key-value pairs.

A **dictionary** in Python is a mutable data structure that stores data as **key-value pairs**.

```python
student = {
    "name": "Vijay",
    "age": 25,
    "course": "Python"
}
```

### Properties of a Dictionary

1. Stores data as **key-value pairs**.
2. Keys must be **unique**.
3. Keys must be **immutable and hashable** types, such as strings, numbers, or tuples.
4. Values can be of **any data type**.
5. Dictionaries are **mutable**, so items can be added, changed, or removed.
6. Dictionaries preserve **insertion order** in modern Python versions.
7. Dictionaries can contain **different data types**.
8. Dictionaries can be **nested** inside other dictionaries.
9. Values may be duplicated, but keys cannot.
10. Dictionaries are **dynamically sized**.
11. They support fast lookup using keys, typically with average $O(1)$ time complexity.
12. An empty dictionary is created using `{}` or `dict()`.
13. Accessing a missing key with `dictionary[key]` raises a `KeyError`; `.get()` can avoid this.
14. Dictionaries are iterable, and iteration normally proceeds over their keys.

## 4. How can you merge two dictionaries in Python? 

````python
dict_1 = {"x": 1, "y":2, "z":3}
dict_2 = {"a": 1, "b":2, "c":3}

dict_3 = {**dict_1, **dict_2}

print(dict_3)

## another approach

dict_4 = dict_1 | dict_2
print(dict_4)
````

## 6.How can you iterate over a dictionary to access keys, values, and key-value pairs?

````python
my_dict = {"x": 1, "y":2, "z":3}

## printing keys
for key in my_dict:
    print(key)

## printing values
for value in my_dict.values():
    print(value)

## printing key,value pairs 
for key,value in my_dict.items():
    print(f"{key} : {value}")
````

## 8. Count the frequency of each word in a given sentence using a dictionary. 

````python
def count_words(sentence):
    words_dict = {}
    for word in sentence.split():
        words_dict[word] = words_dict.get(word, 0) + 1

    for key, value in words_dict.items():
        print(f"Word '{key}' is repeated {value} times. \n")


sentence = "the frog jumped over the ball"

count_words(sentence=sentence)

````

## 9. Count the frequency of each character in a given word using a dictionary. 

````python
def count_letters(word):
    letters_dict = {}
    for letter in word:
        letters_dict[letter] = letters_dict.get(letter, 0) + 1

    for key, value in letters_dict.items():
        print(f"Letter '{key}' is repeated {value} times. \n")


word = "unforgettable"

count_letters(word=word)

````

## 10. Sort a dictionary by keys and values. 

````python
my_dict = {"c": 3, "a": 1, "b": 2}

sorted_by_keys = dict(sorted(my_dict.items()))
print(sorted_by_keys)

sorted_by_values_in_asc_order = dict(sorted(my_dict.items(), key=lambda item: item[1]))
sorted_by_values_in_desc_order = dict(sorted(my_dict.items(), key=lambda item: item[1],reverse=True))
print(sorted_by_values_in_asc_order)
print(sorted_by_values_in_desc_order)
````

## 13. Group a list of words into lists of anagrams using a dictionary. Print the list of lists where each
## inner list contains words that are anagrams of each other.
## ● Input: ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
## ● Anagram groups: [['listen', 'silent', 'enlist'], ['hello'], ['lemon', 'melon'], ['debitcard','badcredit']] 

````python
input_list = ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"] 

def generate_hash(word):
    sorted_word = "".join(sorted(word))
    return sorted_word    


def group_anagrams(input_list):
    output_dict = {}

    for word in input_list:
        word_hash = generate_hash(word)
        output_dict.setdefault(word_hash,[]).append(word)

    output_list = [elem for elem in output_dict.values()]
    return output_list



print(group_anagrams(input_list=input_list))




````

## 14. Merge two dictionaries. If there are common keys, sum their values.
## Print the merged dictionary. 

````python
from collections import defaultdict

dict_1 = {"a": 10, "b": 20, "c": 30}
dict_2 = {"b": 5, "c": 15, "d": 40}

def merge_dictionary(dict_1,dict_2):
    merged_dictionary = defaultdict(int)

    for key,value in dict_1.items():
        merged_dictionary[key] += value

    for key,value in dict_2.items():
        merged_dictionary[key] += value

    return dict(merged_dictionary)

print(merge_dictionary(dict_1=dict_1,dict_2=dict_2))            
         
````
